```python
import pandas as pd

# Membaca data mentah masing-masing polutan
co = pd.read_csv("CO_Kota_Sumenep_raw.csv")
no2 = pd.read_csv("NO2_Kota_Sumenep_raw.csv")
so2 = pd.read_csv("SO2_Kota_Sumenep_raw.csv")

print("=== DATA CO ===")
print("Ukuran data :", co.shape)
print("Nama kolom  :", co.columns.tolist())
display(co.head())

print("\n=== DATA NO2 ===")
print("Ukuran data :", no2.shape)
print("Nama kolom  :", no2.columns.tolist())
display(no2.head())

print("\n=== DATA SO2 ===")
print("Ukuran data :", so2.shape)
print("Nama kolom  :", so2.columns.tolist())
display(so2.head())
```

```python
# Mengubah kolom tanggal menjadi datetime untuk pemeriksaan
co["date"] = pd.to_datetime(co["date"])
no2["date"] = pd.to_datetime(no2["date"])
so2["date"] = pd.to_datetime(so2["date"])

datasets = {
    "CO": co,
    "NO2": no2,
    "SO2": so2
}

for nama, df in datasets.items():
    print(f"\n=== PEMERIKSAAN DATA {nama} ===")
    
    print("\nTipe data:")
    print(df.dtypes)
    
    print("\nMissing value:")
    print(df.isnull().sum())
    
    print("\nJumlah baris duplikat:")
    print(df.duplicated().sum())
    
    print("\nRentang tanggal:")
    print("Awal  :", df["date"].min())
    print("Akhir :", df["date"].max())
    
    print("\nJumlah tanggal unik:")
    print(df["date"].nunique())
```

```python
for nama, df in datasets.items():
    print(f"=== {nama} ===")
    print("Tipe polutan     :", df[nama].dtype)
    print("Missing date     :", df["date"].isnull().sum())
    print("Missing polutan  :", df[nama].isnull().sum())
    print("Duplikat         :", df.duplicated().sum())
    print("Tanggal awal     :", df["date"].min())
    print("Tanggal akhir    :", df["date"].max())
    print("Tanggal unik     :", df["date"].nunique())
    print()
```

```python
print("=== STATISTIK DESKRIPTIF DATA MENTAH ===")

print("\n=== CO ===")
print(co["CO"].describe())

print("\n=== NO2 ===")
print(no2["NO2"].describe())

print("\n=== SO2 ===")
print(so2["SO2"].describe())
```

```python
print("=== SO2 ===")
print("count :", so2["SO2"].count())
print("mean  :", so2["SO2"].mean())
print("std   :", so2["SO2"].std())
print("min   :", so2["SO2"].min())
print("25%   :", so2["SO2"].quantile(0.25))
print("50%   :", so2["SO2"].quantile(0.50))
print("75%   :", so2["SO2"].quantile(0.75))
print("max   :", so2["SO2"].max())
```

```python
import matplotlib.pyplot as plt
import numpy as np
import ast

# Salinan data khusus untuk visualisasi
co_plot = co.copy()

# Mengambil nilai numerik CO dari format seperti [0.0257]
def ambil_nilai_co(x):
    try:
        nilai = ast.literal_eval(str(x))

        if isinstance(nilai, list):
            if len(nilai) == 0 or nilai[0] is None:
                return np.nan
            return float(nilai[0])

        return float(nilai)
    except:
        return np.nan

co_plot["CO_numeric"] = co_plot["CO"].apply(ambil_nilai_co)

# Membuat visualisasi
fig, axes = plt.subplots(3, 1, figsize=(14, 12))

# CO
axes[0].plot(co_plot["date"], co_plot["CO_numeric"])
axes[0].set_title("Time Series Karbon Monoksida (CO)")
axes[0].set_xlabel("Tanggal")
axes[0].set_ylabel("CO")
axes[0].grid(True, alpha=0.3)

# NO2
axes[1].plot(no2["date"], no2["NO2"])
axes[1].set_title("Time Series Nitrogen Dioksida (NO₂)")
axes[1].set_xlabel("Tanggal")
axes[1].set_ylabel("NO₂")
axes[1].grid(True, alpha=0.3)

# SO2
axes[2].plot(so2["date"], so2["SO2"])
axes[2].set_title("Time Series Sulfur Dioksida (SO₂)")
axes[2].set_xlabel("Tanggal")
axes[2].set_ylabel("SO₂")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# =========================
# 1. TIME SERIES CO
# =========================
plt.figure(figsize=(12, 5))
plt.plot(co_plot["date"], co_plot["CO_numeric"])
plt.title("Time Series Karbon Monoksida (CO)")
plt.xlabel("Tanggal")
plt.ylabel("CO")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("timeseries_CO.png", dpi=300, bbox_inches="tight")
plt.show()


# =========================
# 2. TIME SERIES NO2
# =========================
plt.figure(figsize=(12, 5))
plt.plot(no2["date"], no2["NO2"])
plt.title("Time Series Nitrogen Dioksida (NO₂)")
plt.xlabel("Tanggal")
plt.ylabel("NO₂")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("timeseries_NO2.png", dpi=300, bbox_inches="tight")
plt.show()


# =========================
# 3. TIME SERIES SO2
# =========================
plt.figure(figsize=(12, 5))
plt.plot(so2["date"], so2["SO2"])
plt.title("Time Series Sulfur Dioksida (SO₂)")
plt.xlabel("Tanggal")
plt.ylabel("SO₂")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("timeseries_SO2.png", dpi=300, bbox_inches="tight")
plt.show()
```
